```python
#6
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15","2017-07-19","2013-06-01","2021-02-10","2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(data)

employees.to_csv("employees.csv", index=False)

print(employees)
```

```python
#6a
print("Shape:", df.shape)
#6b
employees.info()
#6c
print(employees.describe())
#6d
print("First 5 rows:")
print(employees.head(5))
print("Last 3 rows:")
print(employees.tail(3))
#6e
#i
average_salary = employees['Salary'].mean()

print("Average Salary:", average_salary)
#ii
total_bonus = employees["Bonus"].sum()

print("Total Bonus:", total_bonus)
#iii
youngest_age = employees["Age"].min()

print("Youngest Age:", youngest_age)
#iv
highest_rating = employees["Rating"].max()

print("Highest Rating:", highest_rating)
```

```python
#6f
sorted_df = employees.sort_values(by="Salary", ascending=False)

print(sorted_df)
```

```python
#6g
def categorize_rating(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance"] = employees["Rating"].apply(categorize_rating)

print(employees[["Name", "Rating", "Performance"]])
```

```python
#6h
print(employees.isnull())
```

```python
#6i
employees.rename(columns={"Employee_ID": "ID"}, inplace=True)

print(employees)
```

```python
#6j
#i
experienced = employees[employees["Years_of_Experience"] > 5]

print(experienced)
#ii
it_employees = employees[employees["Department"] == "IT"]

print(it_employees)
```

```python
#6k
employees["Tax"] = employees["Salary"] * 0.10

print(employees[["Name", "Salary", "Tax"]])
```
