Q1: Build Your Personalized Knowledge Base: Take your college roll number. Extract its digits. Build a pandas
DataFrame with exactly 6 FAQ entries: 4 fixed entries (given in table below) and other 2 entries constructed from your
own roll number digits as follows:

• Take the LAST TWO DIGITS of your roll number. For each digit d, compute category = ["billing", "account",
"general"][d % 3]. Invent one realistic question+answer+3 keywords per entry that fits the assigned category (e.g.
if d%3 gives "account", write a question like “how do I update my registered mobile number”).

• # Example roll number ...23 -> digits 2, 3

• # digit 2 -> category[2 % 3] = general

• # digit 3 -> category[3 % 3] = billing

Output: Print your final 6-row DataFrame.

```python
import pandas as pd

roll_number = "1024170044"

categories = ["billing", "account", "general"]

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

# Last two digits of roll number
last_two_digits = [int(d) for d in roll_number[-2:]]

personalized_entries = []

# Personalized entry for digit 4
digit = last_two_digits[0]
category = categories[digit % 3]

personalized_entries.append({
    "question": "how do i update my registered mobile number",
    "answer": "Go to Account Settings and update your registered mobile number.",
    "keywords": "mobile update phone",
    "category": category
})

# Personalized entry for second digit 4
digit = last_two_digits[1]
category = categories[digit % 3]

personalized_entries.append({
    "question": "how can i change my registered email address",
    "answer": "Go to Account Settings and update your registered email address.",
    "keywords": "email change account",
    "category": category
})

# Combine fixed + personalized entries
all_entries = fixed_entries + personalized_entries

df = pd.DataFrame(all_entries)

print("\n" + "=" * 70)
print("Q1: FINAL 6-ROW KNOWLEDGE BASE")
print("=" * 70)

print(df.to_string(index=False))
```

Q2: Generate and Score a Hypothesis - Implement a scoring function that takes a query string and returns all matching
entries ranked by confidence

```python
import re

def score_query(query, dataframe):
    """
    Takes a query string and returns all matching FAQ entries
    ranked by confidence score.

    Score = number of query words found in the FAQ keywords.
    """

    query_words = set(re.findall(r'\b\w+\b', query.lower()))

    results = []

    for index, row in dataframe.iterrows():

        keyword_words = set(
            re.findall(r'\b\w+\b', row["keywords"].lower())
        )

        matching_words = query_words.intersection(keyword_words)

        if matching_words:
            score = len(matching_words)

            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "matching_keywords": ", ".join(sorted(matching_words)),
                "confidence": score
            })

    # Rank highest score first
    results.sort(key=lambda x: x["confidence"], reverse=True)

    return results


print("\n" + "=" * 70)
print("Q2: QUERY SCORING")
print("=" * 70)

query = "how can I reset my password"

results = score_query(query, df)

print("Query:", query)
print("\nMatching entries:")

for result in results:
    print("\nQuestion:", result["question"])
    print("Answer:", result["answer"])
    print("Category:", result["category"])
    print("Matching keywords:", result["matching_keywords"])
    print("Confidence:", result["confidence"])
```

Q3: Write a function same_category(category_name, df) that returns all questions belonging to a given category. Call it
using the category of one of the personalized entries from Q1, and print the result.

```python
def same_category(category_name, dataframe):
    """
    Returns all FAQ entries belonging to the given category.
    """
    return dataframe[
        dataframe["category"].str.lower() == category_name.lower()
    ]


print("\n" + "=" * 70)
print("Q3: SAME CATEGORY")
print("=" * 70)

# Category of one personalized entry
personalized_category = personalized_entries[0]["category"]

print("Selected category:", personalized_category)

category_result = same_category(personalized_category, df)

print("\nFAQ entries in this category:")
print(category_result.to_string(index=False))
```

Q6: Modify your Q2 scoring function so that if two or more entries tie for the highest score, it does not silently pick one
— it prints all matching entries instead, so the user can see every equally good match. Demonstrate with one query that
produces a tie (e.g. a query matching both "fee" entries) and one that doesn't.

```python
def score_query_with_ties(query, dataframe):
    """
    Scores FAQ entries and handles ties.

    If multiple entries have the highest score,
    all of them are printed.
    """

    query_words = set(re.findall(r'\b\w+\b', query.lower()))

    results = []

    for index, row in dataframe.iterrows():

        keyword_words = set(
            re.findall(r'\b\w+\b', row["keywords"].lower())
        )

        matching_words = query_words.intersection(keyword_words)

        if matching_words:

            score = len(matching_words)

            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "matching_keywords": ", ".join(sorted(matching_words)),
                "confidence": score
            })

    if not results:
        print("\nNo matching FAQ entries found.")
        return []

    # Highest score first
    results.sort(
        key=lambda x: x["confidence"],
        reverse=True
    )

    highest_score = results[0]["confidence"]

    top_matches = [
        result for result in results
        if result["confidence"] == highest_score
    ]

    if len(top_matches) > 1:

        print("\nTIE DETECTED!")
        print(
            "Multiple entries have the highest "
            f"confidence score of {highest_score}."
        )

        for result in top_matches:
            print("\nQuestion:", result["question"])
            print("Answer:", result["answer"])
            print("Category:", result["category"])
            print("Matching keywords:",
                  result["matching_keywords"])
            print("Confidence:", result["confidence"])

    else:

        print("\nUnique highest-confidence match:")

        result = top_matches[0]

        print("\nQuestion:", result["question"])
        print("Answer:", result["answer"])
        print("Category:", result["category"])
        print("Matching keywords:",
              result["matching_keywords"])
        print("Confidence:", result["confidence"])

    return results


# ------------------------------------------------------------
# Q6 Demonstration 1: Query that produces a tie
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("Q6: TIE DEMONSTRATION")
print("=" * 70)

tie_query = "fee"

print("\nQuery:", tie_query)

score_query_with_ties(tie_query, df)


# ------------------------------------------------------------
# Q6 Demonstration 2: Query without a tie
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("Q6: NON-TIE DEMONSTRATION")
print("=" * 70)

non_tie_query = "password reset"

print("\nQuery:", non_tie_query)

score_query_with_ties(non_tie_query, df)
```
