```python
#3 (i)
import random
print("a list of 100 random numbers between 100 and 900:")
random_list=[random.randint(100,900) for i in range(100)]
print(random_list)
```

```python
#3 (ii)
print("odd numbers in random list:")
odd_numbers = [num for num in random_list if num % 2 != 0]
print(odd_numbers)
print(len(odd_numbers))
```

```python
#3 (iii)
even_numbers = [num for num in random_list if num % 2 == 0]
print(even_numbers)
print(len(even_numbers))
```

```python
#3 (iv)
from sympy import isprime
primes = [x for x in random_list if isprime(x)]
print(primes)
print(len(primes))
```

```python
#3 (v)
most_common = max(set(random_list), key=random_list.count)
print(most_common)
print(random_list.count(most_common))
```

```python
A=[int(digit)*7 for digit in rno]
B=[int(digit)*9 for digit in rno]
print(A)
print(B)

#4 (vi)
union = list(set().union(A, B))
print(union)

#4 (vii)
intersection = list(set(A) & set(B))
print(intersection)

#4 (viii)
# A − B
print(set(A).difference(B))

# B − A
print(set(B).difference(A))

#4 (ix)
print(set(A).symmetric_difference(B))

#4 (x)
# Check whether A is a subset of B
print(set(A).issubset(B))

# Check whether B is a superset of A
print(set(B).issuperset(A))

#4 (xi)

X = int(input("Enter a value to remove: "))

set(A).discard(X)

print("Set A after removing X:", A)

#discard() does nothing if X is not present, whereas remove() raises a KeyError
```

```python
#5
# Original dictionary
my_dict = {
    "name": "Samuel Masih",
    "roll_no": "YOUR_ROLL_NO",
    "branch": "Computer Engineering",
    "age": 20,
    "city": "Jalandhar"
}

# i. Rename "city" to "location" using pop()
my_dict["location"] = my_dict.pop("city")

# ii. Add CGPA
my_dict["cgpa"] = 8.5

# iii. Increase age by 1
my_dict["age"] += 1

print("Updated dictionary:")
print(my_dict)


# iv. Delete "branch" using pop() and del in two separate copies

# Copy 1: using pop()
dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")

# Copy 2: using del
dict_del = my_dict.copy()
del dict_del["branch"]

print("\nUsing pop():")
print(dict_pop)

print("\nUsing del:")
print(dict_del)

# pop() removes the key and returns its value, while del removes the key
# without returning its value.


# v. Iterate using .items()
print("\nKey-value pairs:")
for key, value in my_dict.items():
    print(f"{key} → {value}")


# vi. Check whether "email" exists before accessing it
if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail key does not exist.")


# vii. Create friend's dictionary
friend_dict = {
    "name": "Rahul Sharma",
    "roll_no": "CE2026001",
    "branch": "Computer Engineering",
    "age": 21,
    "city": "Delhi"
}

# Merge using dictionary unpacking
merged_dict = {**my_dict, **friend_dict}

print("\nMerged dictionary:")
print(merged_dict)

# If both dictionaries have the same key, values from the second
# dictionary (friend_dict) overwrite the values from the first dictionary.


# viii. Dictionary comprehension: only key-value pairs where value is a string
string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nDictionary containing only string values:")
print(string_values)
```
