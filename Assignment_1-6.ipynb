## Assignment 5.2
**WAP to add all numbers divisible by 7 and 9 from 1 to n, where n is given by the user.**

```python
n = int(input("Enter n: "))

total = 0

for i in range(1, n + 1):
    if i % 7 == 0 and i % 9 == 0:
        total = total + i

print("Sum =", total)
```

## Assignment 5.3
**WAP to add all prime numbers from 1 to n, where n is given by the user.**

```python
n = int(input("Enter n: "))

total = 0

for num in range(2, n + 1):
    is_prime = True

    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        total = total + num

print("Sum of prime numbers =", total)
```

## Assignment 6.1
**WAP using a function that adds all odd numbers from 1 to n, where n is given by the user.**

```python
def sum_odd_numbers(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            total = total + i

    return total


n = int(input("Enter n: "))

print("Sum of odd numbers =", sum_odd_numbers(n))
```

## Assignment 6.2
**WAP using a function that adds all prime numbers from 1 to n, where n is given by the user.**

```python
def is_prime(num):
    if num < 2:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True


def sum_prime_numbers(n):
    total = 0

    for i in range(1, n + 1):
        if is_prime(i):
            total = total + i

    return total


n = int(input("Enter n: "))

print("Sum of prime numbers =", sum_prime_numbers(n))
```
